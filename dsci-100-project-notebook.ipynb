#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
wmm_summary_stats <- tidy_women_mm |>
    mutate(mean_conf_wins = mean(conf_wins),
          max_conf_wins = max(conf_wins),
          min_conf_wins = min(conf_wins)) |>
    mutate(mean_conf_losses = mean(conf_losses),
          max_conf_losses = max(conf_losses),
          min_conf_losses = mean(conf_losses)) |>
    mutate(mean_conf_rank = mean(conf_rank),
          max_conf_rank = max(conf_rank),
          min_conf_rank = min(conf_rank)) |>
    select(c(mean_conf_wins:min_conf_rank)) |>
    slice(1)

wmm_summary_stats
```

```r
options(repr.plot.width = 8, repr.plot.height = 5)
plot_1 <- tidy_women_mm |>
    ggplot(aes(x = conf_losses)) +
    geom_histogram(bins = 15) +
    xlab("Number of Conference Losses") +
    ylab("Frequency") +
    theme(text = element_text(size = 15))

plot_1
```

```r
options(repr.plot.width = 10, repr.plot.height = 5)
plot_2 <- tidy_women_mm |>
    ggplot(aes(x = conf_wins, fill = bid)) +
    geom_bar(stat = "count") +
    xlab("Conference Ranking") +
    ylab("Count") +
    theme(text = element_text(size = 15))

plot_2
```

In the figure 1 below, it is a scatter plot suggesting a weak negative relationship between number of conference losses and number of conference wins, colored by bid types. These are predictor variables, and it is important to learn the relationship including the bid type. However, many points are overlapped as seen with lower transparency. However, the colored dots are still clustered together, hard to identify, thus suggesting no effect on the response variable. The data is not scaled, so one may dominate over another.

```r
options(repr.plot.width = 8, repr.plot.height = 5)

figure_1 <- tidy_women_mm |>
    ggplot(aes(x = conf_wins, y = conf_losses, color = bid)) +
    geom_point(alpha = 0.2, size = 1.5) +
    labs(x = "Number of Conference Wins",
        y = "Number of Conference Losses",
        color = "Bids") +
    theme(text = element_text(size = 15))

figure_1
```

This is relevant as not only they are predictor variables and we need to understand their relationship, but also we need to understand them by including the bid types. By looking at colored dots with low transparency, we can see that bright blue dots are clustered on the left side, and bright red are in the middle, suggesting an effect on response variables. The data is not scaled, so one may dominate over another.

```r
options(repr.plot.width = 8, repr.plot.height = 5)

figure_2 <- tidy_women_mm |>
    ggplot(aes(x = conf_rank, y = conf_wins, color = bid)) +
    geom_point(alpha = 0.2, size = 1.5) +
    labs(x = "Place in Conference",
        y = "Number of Conference Wins",
        color = "Bids") +
    theme(text = element_text(size = 15))

figure_2
```
